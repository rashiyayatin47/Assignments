```python
# Write a recursive function in Python called reverse_string(s) that takes a string and returns 
# it reversed (e.g., 'hello' becomes 'olleh').

def reverse(a):
    if a == "":
        return ""
    return reverse(a[1:]) + a[0]

reverse("hello")
```

```python
# Build a recursive function sum_playlist_durations(durations) that takes a list of song durations
#  (in seconds) and returns the total duration, similar to how Spotify totals a playlist.

def sum_playlist(durations):
    if len(durations) == 0:
        return 0
    return durations[0] + sum_playlist(durations[1:])

songs = [180, 240, 150, 300]

print(sum_playlist(songs))
```

```python
# Create a recursive function count_likes(posts) that takes a nested dictionary representing Instagram 
# posts and their replies (each with a 'likes' key), and returns the total number of likes across all
# posts and replies.<br><br><em><strong>Hint:</strong> Each reply can itself have more replies,
#  so use recursion to sum likes at all levels.</em>


def count_likes(post):
    total = post["likes"]

    for reply in post["replies"]:
        total += count_likes(reply)

    return total


posts = {"likes": 10,"replies": [ { "likes": 5,"replies":[] },{"likes": 8,"replies": [{"likes": 2,"replies": []}]}]}

print("Total Likes:", count_likes(posts))
```
