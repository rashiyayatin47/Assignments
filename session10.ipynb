#Task 2

```python
# Given a list of song titles from Spotify with extra spaces and inconsistent casing, use map() and a lambda function to clean each title so that it is stripped of spaces and converted to title case (e.g., ' shape OF you ' → 'Shape Of You').
songs = [" shape OF you "," blinding LIGHTS "," kesariya "," believer "]

clean_songs = list(map(lambda song: song.strip().title(), songs))

print(clean_songs)
```

#Task 3

```python
# Use filter() and a lambda function to extract only those Flipkart product names from a list that start
#  with the letter 'S' (case-insensitive).
products = ["Samsung Galaxy S25","iPhone 16","sony Headphones","Realme Narzo","Smart Watch"]

s_products = list(filter(lambda p: p.lower().startswith('s'), products))

print(s_products)
```

#Task 4

```python
# Given a list of order amounts from a Zomato cart [120, 340, 560, 80], use reduce() from functools to
#  calculate the total bill amount.

from functools import reduce

orders = [120, 340, 560, 80]

total_bill = reduce(lambda x, y: x + y, orders)

print("Total Bill:", total_bill)
```

```python
# Use ChatGPT or Copilot to generate a Python code snippet that uses map(), filter(), and reduce() 
# together to process a list of numbers: first double each number, then filter to keep only numbers
#  greater than 100, and finally sum the result. Paste and test the generated code with the list
#  [40, 60, 80, 120].
from functools import reduce

numbers = [40, 60, 80, 120]

doubled = map(lambda x: x * 2, numbers)
filtered = filter(lambda x: x > 100, doubled)
result = reduce(lambda x, y: x + y, filtered)

print(result)
```
